```python
import requests
from bs4 import BeautifulSoup
```

```python
url="https://www.flipkart.com/apple-iphone-13-starlight-128-gb/p/itmc9604f122ae7f?pid=MOBG6VF5ADKHKXFX&lid=LSTMOBG6VF5ADKHKXFX4LCPEV&marketplace=FLIPKART&store=tyy%2F4io&srno=b_1_1&otracker=browse&fm=organic&iid=1d0ea575-85c0-4b7d-97a5-cb3efc618a42.MOBG6VF5ADKHKXFX.SEARCH&ppt=hp&ppn=homepage&ssid=g43gteannk0000001746097682127"
headers={"user_agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36 Edg/135.0.0.0"}
req=requests.get(url,headers=headers)
soup=BeautifulSoup(req.content,"html.parser")
```

```python
reviews =soup.findAll('div',{'class':'ZmyHeo'});
for review in reviews:
    print(review.get_text()+"\n\n")
```

```python
ratings=soup.find('div',{'class':'ipqd2A'}).get_text()
print(ratings)
```

```python
individual_ratings=soup.findAll('div',{'class':'XQDdHH Ga3i8K'})
for ratings in individual_ratings:
    print(ratings.get_text())

                                       
```
