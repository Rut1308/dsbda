```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
data = pd.read_csv('iris.csv')
```

```python
data.head()
```

```python
data.hist(figsize=(10,10),bins=15,edgecolor='black')
plt.suptitle('Histogram')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=data.drop(columns='species'))
plt.title('boxplot')
plt.show()
```

```python
plt.figure(figsize=(10,8))
data.groupby('species').mean().T.plot(marker='o')
plt.title("Single Line Graph: Mean Feature Values per Species")
plt.ylabel("Mean Value")
plt.xlabel("Features")
plt.grid(True)
plt.legend(title="species")
plt.show()
           
```

```python
plt.figure(figsize=(10,8))
plt.plot(data["sepal_length"],label="sepal_length")
plt.plot(data["sepal_width"],label="sepal_width")
plt.plot(data["petal_length"],label="petal_length")
plt.plot(data["petal_width"],label="petal_width")
plt.title("Multiple Line Graph of Features")
plt.xlabel("Index")
plt.ylabel("Measurement")
plt.legend()
plt.grid(True)
plt.show()
```
